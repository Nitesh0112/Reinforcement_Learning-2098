```python
# Value Iteration in a Deterministic 2x2 Grid World

gamma = 0.9
theta = 0.001

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

# Transition rules
transitions = {
    'S0': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S0',
        'Right': 'S0'
    },
    'S1': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S1',
        'Right': 'G'
    }
}

# Initialize state values
V = {'S0': 0, 'S1': 0, 'G': 0}

# Rewards
goal_reward = 100
normal_reward = -1

iteration = 0

# Value Iteration
while True:
    delta = 0
    new_V = V.copy()

    for state in ['S0', 'S1']:
        Q = []

        for action in actions:
            next_state = transitions[state][action]

            if next_state == 'G':
                reward = goal_reward
            else:
                reward = normal_reward

            value = reward + gamma * V[next_state]
            Q.append(value)

        new_V[state] = max(Q)
        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V
    iteration += 1

    print(f"Iteration {iteration}: S0 = {V['S0']:.2f}, S1 = {V['S1']:.2f}")

    if delta < theta:
        break

# Calculate final Q-values
print("\nFinal Action Values:")

for state in ['S0', 'S1']:
    print(f"\nState: {state}")

    for action in actions:
        next_state = transitions[state][action]

        reward = goal_reward if next_state == 'G' else normal_reward

        Q = reward + gamma * V[next_state]

        print(f"{action}: {Q:.2f}")

# Display optimal policy
print("\nOptimal Values:")
for state in ['S0', 'S1']:
    print(state, "=", round(V[state], 2))

print("G = 0")

print("\nOptimal Policy:")
for state in ['S0', 'S1']:
    best_action = max(
        actions,
        key=lambda a: (
            (goal_reward if transitions[state][a] == 'G' else normal_reward)
            + gamma * V[transitions[state][a]]
        )
    )
    print(state, "->", best_action)
```
