```python
# Assignment 1: Calculate Undiscounted and Discounted Returns

n = int(input("Enter number of time steps: "))

rewards = []

for i in range(n):
    reward = float(input(f"Enter reward at time step {i + 1}: "))
    rewards.append(reward)

# Undiscounted return
undiscounted_return = sum(rewards)

# Discount factor
gamma = float(input("Enter discount factor (gamma): "))

# Discounted return
discounted_return = 0

for i in range(n):
    discounted_return += (gamma ** i) * rewards[i]

print("\nEntered reward values:", rewards)
print("Undiscounted return:", undiscounted_return)
print("Discounted return:", discounted_return)
```

```python
# Assignment 2: 2 x 3 Grid World MDP

# State Space
states = [
    (0, 0), (0, 1), (0, 2),
    (1, 0), (1, 1), (1, 2)
]

# Action Space
actions = ["up", "down", "left", "right"]

# Goal and Obstacle
goal = (0, 2)
obstacle = (1, 1)

# Discount Factor
gamma = 0.9


# Reward Function
def reward(state):
    if state == goal:
        return 10
    elif state == obstacle:
        return -10
    else:
        return -1


# Transition Function
def transition(state, action):

    row, col = state

    if action == "up":
        row -= 1
    elif action == "down":
        row += 1
    elif action == "left":
        col -= 1
    elif action == "right":
        col += 1

    # Check boundary
    if row < 0 or row > 1 or col < 0 or col > 2:
        return state

    # Check obstacle
    if (row, col) == obstacle:
        return state

    return (row, col)


# Display MDP
print("State Space:", states)
print("Action Space:", actions)
print("Discount Factor:", gamma)

# User Input
row = int(input("\nEnter current state row (0 or 1): "))
col = int(input("Enter current state column (0, 1 or 2): "))
action = input("Enter action (up/down/left/right): ").lower()

current_state = (row, col)

# Calculate next state
next_state = transition(current_state, action)

# Display result
print("\nCurrent State:", current_state)
print("Action:", action)
print("Next State:", next_state)
print("Reward:", reward(next_state))
```
