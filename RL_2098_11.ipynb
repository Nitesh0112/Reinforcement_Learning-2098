```python
# Policy Iteration in Deterministic Grid World

gamma = 0.9
theta = 0.001

states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

# Initial policy
policy = {
    'S0': 'Up',
    'S1': 'Up'
}

# Initialize state values
V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

# Transition function
def transition(state, action):

    if state == 'S0':
        moves = {
            'Up': ('S0', -1),
            'Down': ('S1', -1),
            'Left': ('S0', -1),
            'Right': ('S0', -1)
        }

    elif state == 'S1':
        moves = {
            'Up': ('S0', -1),
            'Down': ('S1', -1),
            'Left': ('S1', -1),
            'Right': ('G', 100)
        }

    return moves[action]


# Policy Iteration
iteration = 0

while True:

    iteration += 1

    # Policy Evaluation
    while True:
        delta = 0

        for state in states:
            old_value = V[state]

            next_state, reward = transition(
                state, policy[state]
            )

            V[state] = reward + gamma * V[next_state]

            delta = max(delta, abs(old_value - V[state]))

        if delta < theta:
            break

    print("\nIteration:", iteration)
    print("State Values:", V)

    # Policy Improvement
    policy_stable = True

    for state in states:

        old_action = policy[state]

        q_values = {}

        for action in actions:

            next_state, reward = transition(state, action)

            q_values[action] = (
                reward + gamma * V[next_state]
            )

        best_action = max(q_values, key=q_values.get)

        policy[state] = best_action

        if old_action != best_action:
            policy_stable = False

    if policy_stable:
        break


# Final Results
print("\n----- FINAL OPTIMAL RESULTS -----")

for state in states:
    print(state, "Value:", round(V[state], 2),
          "Optimal Action:", policy[state])

print("G Value:", V['G'])


# Action Value Table
print("\n----- Q VALUES -----")

for state in states:

    print("\n", state)

    for action in actions:

        next_state, reward = transition(state, action)

        q = reward + gamma * V[next_state]

        print(action, ":", round(q, 2))


# Bellman Optimality Verification
print("\n----- BELLMAN OPTIMALITY -----")

for state in states:

    q_values = []

    for action in actions:

        next_state, reward = transition(state, action)

        q = reward + gamma * V[next_state]

        q_values.append(q)

    print(state,
          "V* =", round(V[state], 2),
          "Max Q =", round(max(q_values), 2))
```
