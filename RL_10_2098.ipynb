```python
# Practical 10: Bellman Optimality and Value Iteration

# 1. Define states and actions
states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.001

# 2. Define transitions
transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S0', -1),
        'Right': ('S0', -1)
    },
    'S1': {
        'Up': ('S0', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 10)
    }
}

# 3. Display MDP information
print("STATES:", states)
print("ACTIONS:", actions)
print("Discount Factor:", gamma)

print("\nTRANSITION DETAILS")

for state in ['S0', 'S1']:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state, "Reward:", reward)

# 4. Initialize state values
V = {'S0': 0, 'S1': 0, 'G': 0}

print("\nINITIAL VALUES:", V)

# 5. Value Iteration
iteration = 0

while True:
    new_V = V.copy()
    delta = 0

    for state in ['S0', 'S1']:
        values = []

        for action in actions:
            next_state, reward = transitions[state][action]

            value = reward + gamma * V[next_state]
            values.append(value)

        new_V[state] = max(values)

    for state in ['S0', 'S1']:
        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V
    iteration += 1

    print("\nIteration:", iteration)
    print("State Values:", V)
    print("Delta:", round(delta, 4))

    if delta < theta:
        break

# 6. Calculate Q-values
print("\nFINAL OPTIMAL STATE VALUES")
print(V)

print("\nACTION VALUES (Q*)")

Q = {}

for state in ['S0', 'S1']:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]

        q_value = reward + gamma * V[next_state]
        Q[state][action] = q_value

        print(state, action, ":", round(q_value, 4))

# 7. Extract optimal policy
print("\nOPTIMAL POLICY")

for state in ['S0', 'S1']:
    max_value = max(Q[state].values())

    best_actions = [
        action for action, value in Q[state].items()
        if abs(value - max_value) < 1e-9
    ]

    print(state, ":", best_actions)

# 8. Verify Bellman Optimality
print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in ['S0', 'S1']:
    max_q = max(Q[state].values())

    print(
        state,
        "V*(s) =", round(V[state], 4),
        "Max Q*(s,a) =", round(max_q, 4)
    )

    if abs(V[state] - max_q) < 1e-9:
        print("Verification Successful")
    else:
        print("Verification Failed")

# 9. Explanation
print("\nEXPLANATION")
print("The optimal policy selects actions with maximum Q-values.")
print("The goal reward of +10 increases the value of nearby states.")
print("The discount factor gamma = 0.9 determines future reward importance.")
print("Value Iteration stops when delta becomes smaller than theta.")
```
