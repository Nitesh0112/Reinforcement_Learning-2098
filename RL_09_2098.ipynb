Assignment 1: Bellman Equations (Simple Python Code)

```python
gamma = 0.9

# Solve state values algebraically
# V0 = 0.9 * (-1 + gamma * V1) + 0.1 * (-1)
# V1 = 0.9 * 100 + 0.1 * (-1 + gamma * V0)

V0 = 98.0 / 1.081
V1 = 89.9 + 0.09 * V0

# Calculate action values
Q0_down = -1 + gamma * V1
Q0_right = -1

Q1_right = 100 * 0.9 + 0.1 * (-1 + gamma * V0)
Q1_up = -1 + gamma * V0

print("V(S0) =", round(V0, 2))
print("V(S1) =", round(V1, 2))
print("Q(S0, Down) =", round(Q0_down, 2))
print("Q(S0, Right) =", round(Q0_right, 2))
print("Q(S1, Right) =", round(Q1_right, 2))
print("Q(S1, Up) =", round(Q1_up, 2))
```

Assignment 2: Iterative Policy Evaluation

```python
gamma = 0.9
theta = 0.000001

V0 = 0
V1 = 0
iterations = 0

while True:
    # Calculate new state values
    new_V0 = 0.1 * (-1) + 0.9 * (-1 + gamma * V1)
    new_V1 = 0.9 * 100 + 0.1 * (-1 + gamma * V0)

    # Calculate difference
    delta = max(abs(new_V0 - V0), abs(new_V1 - V1))

    # Update values
    V0 = new_V0
    V1 = new_V1
    iterations += 1

    # Stop when values converge
    if delta < theta:
        break

# Calculate action values
Q0_down = 0.1 * (-1) + 0.9 * (-1 + gamma * V1)
Q0_right = -1
Q1_right = 0.9 * 100 + 0.1 * (-1 + gamma * V0)
Q1_up = -1 + gamma * V0

# Display results
print("Assignment 2 Results")
print("--------------------")
print("Iterations =", iterations)
print("V(S0) =", round(V0, 2))
print("V(S1) =", round(V1, 2))
print("Q(S0, Down) =", round(Q0_down, 2))
print("Q(S0, Right) =", round(Q0_right, 2))
print("Q(S1, Right) =", round(Q1_right, 2))
print("Q(S1, Up) =", round(Q1_up, 2))

# Verify policy consistency
print("Policy Check S0:", abs(V0 - Q0_down) < theta)
print("Policy Check S1:", abs(V1 - Q1_right) < theta)
```
