```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of machines
M = 3

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# Initialize N, R and Q tables
N = {'A': 0, 'B': 0, 'C': 0}   # Number of times played
R = {'A': [], 'B': [], 'C': []} # Rewards received
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0} # Estimated values

total_reward = 0

# Part C - Epsilon-Greedy Decision
for play in range(1, N_play + 1):

    r = random.random()

    if r < epsilon:
        # Exploration
        selected_machine = random.choice(machines)
        action = "Exploration"
    else:
        # Exploitation
        max_q = max(Q.values())

        # Choose randomly if multiple machines have same Q-value
        best_machines = [m for m in machines if Q[m] == max_q]
        selected_machine = random.choice(best_machines)
        action = "Exploitation"

    # Part D - Reward Collection
    print(f"\nPlay {play}")
    print(f"Selected Machine: {selected_machine}")
    print(f"Strategy: {action}")

    reward = float(input("Enter reward: "))

    # Update number of times selected
    N[selected_machine] += 1

    # Store reward
    R[selected_machine].append(reward)

    # Part E - Incremental Sample-Average Update
    Q[selected_machine] = Q[selected_machine] + (
        (reward - Q[selected_machine]) / N[selected_machine]
    )

    # Update total reward
    total_reward += reward

    print(f"Updated Q-value of {selected_machine}: "
          f"{Q[selected_machine]:.2f}")

# Part F - Display Results
print("\n" + "=" * 45)
print("FINAL RESULTS")
print("=" * 45)

print("\nMachine\tN\tRewards\t\tQ-Value")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]}\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward after 10 plays:", total_reward)

# Best machine
best_machine = max(Q, key=Q.get)
print("Best Machine:", best_machine)
print("Highest Q-value:", round(Q[best_machine], 2))
```
