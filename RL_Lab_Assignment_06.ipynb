```python
import gymnasium as gym

# Function to run Random Agent
def run_random_agent(env_name, episodes=5):

    print("\n========================================")
    print("Environment:", env_name)
    print("========================================")

    # 1. Create environment
    env = gym.make(env_name)

    episode_rewards = []

    # Repeat for each episode
    for episode in range(1, episodes + 1):

        # 2. Reset environment
        state, info = env.reset()

        total_reward = 0
        steps = 0

        terminated = False
        truncated = False

        # Continue until episode terminates
        while not (terminated or truncated):

            # 3. Observe current state
            # print("Current State:", state)

            # 4. Identify available actions
            # 5. Select an action randomly
            action = env.action_space.sample()

            # 6. Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # 7. Observe next state and reward
            state = next_state

            # Add reward
            total_reward += reward
            steps += 1

        # 8. Episode terminated

        # 9. Calculate average reward
        average_reward = total_reward / steps if steps > 0 else 0

        episode_rewards.append(total_reward)

        # Display results
        print("\nEpisode:", episode)
        print("Steps:", steps)
        print("Total Reward:", total_reward)
        print("Average Reward:", round(average_reward, 2))

    # 10. Print total reward for each episode
    print("\nTotal rewards for all episodes:")
    print(episode_rewards)

    env.close()


# ========================================
# Run Random Agent on CliffWalking
# ========================================
run_random_agent("CliffWalking-v1", episodes=5)


# ========================================
# Run Random Agent on Taxi
# ========================================
run_random_agent("Taxi-v4", episodes=5)
```
